---

The aim of this exercise is to immerse students in the **newsvendor problem**, a classic single-period inventory decision-making scenario. Students must determine an optimal order quantity when faced with **uncertain demand** and the trade-off between the risks of:

- **Overstocking** (leading to salvage losses), and
- **Understocking** (leading to missed sales opportunities).

Through repeated rounds of ordering decisions, students gain hands-on experience with:

1. **Decision-making under risk** — balancing surplus and shortage consequences when demand is uncertain.
2. **Economic utility theory** — understanding the rational, risk-neutral approach to finding optimal order quantities.
3. **Behavioral decision-making** — recognizing how cognitive biases, bounded rationality, and risk perception can lead to suboptimal choices.
4. **Operational context** — appreciating how these dynamics still apply to modern supply chains, particularly those dealing with perishable goods or seasonal products.
5. **Prospect Theory connection** — understanding how perceptions of gains and losses, rather than pure expected value, shape ordering decisions in predictable ways.

# 📰 Newsvendor Game — Instructions

You will play a simulation game that represents a **single-period inventory decision problem**.
Your goal is to **maximize your total profit** over multiple rounds.

---

## 🎮 How to Play

### 1. Starting the Game

You will play **two separate scenarios**, each with different business conditions.

**Scenario A — Valentine's Roses**
You are a florist preparing for **Valentine's Day**. Roses are in huge demand, but only for one day.
Any roses you don't sell on February 14th lose most of their value the next day.
If you order too many, you risk having unsold flowers that must be sold at a steep discount.
If you order too few, you miss the chance to sell more while customers are willing to pay full price.
- **Parameters:** selling price p = $12, purchase cost c = $3, salvage value h = $0.
- **Example:** You order **550 roses** and actual demand is **525 roses**:
  - You sell 525 roses at the full selling price ($12 each).
  - You have 25 unsold roses, which are worth $0 the next day.
  - Your profit = sales revenue − cost of roses ordered + salvage value of leftovers.
  - You also see the *opportunity cost* — profit you could have made if your order matched demand exactly.

**Scenario B — Daily Newspaper Stand**
You run a small newspaper stand. Each morning, you place your order before knowing how many customers will buy that day.
Unsold newspapers at the end of the day have very little value.
Ordering too many means paying for papers you can't sell.
Ordering too few means turning away customers — and missing those sales entirely.
- **Parameters:** selling price p = $12, purchase cost c = $9, salvage value h = $0.
- **Example:** You order **550 copies** and actual demand is **575 copies**:
  - You sell all 550 copies you have ($12 each).
  - 25 customers leave without buying — missed sales.
  - Your profit = revenue from sales − cost of papers purchased.
  - The *opportunity cost* here is the lost profit from not stocking enough copies.

Click **Start / Reset** to begin a scenario.

### 2. Making Decisions

- Each round, you will **choose an order quantity (q)** before knowing the demand.
- Demand will vary between **500 and 600 units**, in increments of 25 — but you are **not told the underlying distribution**. Part of your job is to *learn it* from what you observe.
- After each round, you will see:
  - The realized demand
  - Your profit for the round
  - Losses from overstocking (salvage loss) or understocking (shortage loss)
  - **Your full history of realized demand so far**, including a running mean and standard deviation, and two charts (demand-by-round and demand frequency) — see "Historical Demand Data" below the decision tables.
- **You cannot carry over unsold goods** to the next round.

### 3. Game Length

- You will play **at least 20 iterations (rounds)** for each scenario.
- The **Rounds** field defaults to 20, but you can raise it (up to 200) at any
  time, even mid-game, if your mu-hat/sigma-hat estimate hasn't stabilized yet.
  More rounds = a more reliable estimate, at the cost of more time playing.

### 4. After the Game

- Click **Download CSV** to save your round-by-round results.
- The CSV file contains your decisions, outcomes, and running demand statistics for every round.

## 📊 Using the Historical Demand Data to Estimate Q\*

As you play, the game keeps a running tally of every demand value you've observed. Below the decision tables you'll find:

- A **table** of realized demand by round, with the running **mean (μ̂)** and **standard deviation (σ̂)** updated after every round.
- A **line chart** showing demand realized each round, with your running mean overlaid, so you can see how your estimate stabilizes as you collect more data.
- A **frequency chart** showing how often each demand value (500, 525, 550, 575, 600) has occurred so far.

**Why this matters:** the classic newsvendor solution says you should order up to the point where the *critical ratio*

**CR = Cu / (Cu + Co)**

is matched by the cumulative probability of demand, where Cu = p - c (underage cost — profit lost per unit of unmet demand) and Co = c - h (overage cost — loss per unit left unsold). If you're willing to approximate demand as roughly bell-shaped, this becomes:

**Q\* ≈ μ̂ + z_CR × σ̂**

where z_CR is the z-score corresponding to your critical ratio. **You now have everything you need to estimate μ̂ and σ̂ yourself from the historical data table/charts** — we'll formalize the exact z-score lookup and discuss the discrete vs. normal-approximation nuance in class, but you're encouraged to start estimating now as you play.

> ⚠️ Note: the actual demand in this game is drawn from a small discrete set (500–600, step 25), not a true normal distribution — so your normal-approximation Q\* may not land exactly on one of the five order values. When that happens, round to the nearest allowed order quantity and note that you had to round.

```python
# @title
# Newsvendor Game — Colab ipywidgets version (no extra installs)

import pandas as pd
import numpy as np
import random
from dataclasses import dataclass, asdict
from typing import Tuple, Optional
from IPython.display import display, HTML, clear_output
import ipywidgets as w
import matplotlib.pyplot as plt
import os, datetime
try:
    from google.colab import files
    ON_COLAB = True
except Exception:
    ON_COLAB = False

# --------------------------- Core parameters & math ---------------------------

DEMAND_VALUES = [500, 525, 550, 575, 600]

@dataclass
class Params:
    p: float  # selling price
    c: float  # unit cost
    h: float  # salvage value

@dataclass
class RoundResult:
    round_num: int
    order_q: int
    demand_x: int
    profit: float
    shortage_loss: float
    salvage_loss: float
    cum_profit: float
    running_mean: float
    running_sd: Optional[float]

def profit_and_costs(q: int, x: int, params: Params) -> Tuple[float, float, float]:
    p, c, h = params.p, params.c, params.h
    shortage = max(x - q, 0)
    surplus  = max(q - x, 0)
    shortage_loss = (p - c) * shortage
    salvage_loss  = (c - h) * surplus
    profit = (p * x - c * q + h * (q - x)) if x < q else (p - c) * q
    return profit, shortage_loss, salvage_loss

def critical_ratio(params: Params) -> float:
    Cu = params.p - params.c
    Co = params.c - params.h
    return Cu / (Cu + Co) if (Cu + Co) > 0 else 0.0

def optimal_q_star(params: Params) -> int:
    # NOTE: this uses the TRUE discrete demand distribution (instructor answer key).
    # It is intentionally never shown to students during play; students are asked
    # to estimate Q* themselves from their own observed mean/SD (see markdown above).
    cr = critical_ratio(params)
    n = len(DEMAND_VALUES)
    for i, d in enumerate(DEMAND_VALUES, start=1):
        if (i / n) >= cr: return d
    return DEMAND_VALUES[-1]

def build_profit_table_df(params: Params) -> pd.DataFrame:
    data = []
    for q in DEMAND_VALUES:
        row = []
        for d in DEMAND_VALUES:
            prof, _, _ = profit_and_costs(q, d, params)
            row.append(int(prof))
        data.append(row)
    df = pd.DataFrame(data, index=DEMAND_VALUES, columns=DEMAND_VALUES)
    df.index.name = "Order q"
    return df

def build_opportunity_cost_table_df(params: Params) -> pd.DataFrame:
    p, c, h = params.p, params.c, params.h
    data = []
    for q in DEMAND_VALUES:
        row = []
        for d in DEMAND_VALUES:
            shortage = max(d - q, 0)
            surplus  = max(q - d, 0)
            oc = -((p - c) * shortage + (c - h) * surplus)
            row.append(int(oc))
        data.append(row)
    df = pd.DataFrame(data, index=DEMAND_VALUES, columns=DEMAND_VALUES)
    df.index.name = "Order q"
    return df

# ---------- HTML renderer for column highlighting ----------
def df_to_html_table(df: pd.DataFrame, title: str, highlight_col: Optional[int] = None) -> str:
    styles = """
    <style>
      table.nv {border-collapse: collapse; font-family: ui-sans-serif, system-ui, sans-serif; font-size: 14px;}
      table.nv th, table.nv td {border:1px solid #ccc; padding:6px 10px; text-align: right;}
      table.nv th:first-child, table.nv td:first-child {text-align: left;}
      table.nv th.hi, table.nv td.hi {background:#ffe6e6;}
      caption {text-align:left; font-weight:600; margin:6px 0;}
    </style>
    """
    cols = list(df.columns)
    head_cells = ['<th></th>']
    for c in cols:
        cls = "hi" if (highlight_col is not None and c == highlight_col) else ""
        head_cells.append(f'<th class="{cls}">{c}</th>')
    head_html = "<tr>" + "".join(head_cells) + "</tr>"
    body_rows = []
    for idx, row in df.iterrows():
        cells = [f"<td>{idx}</td>"]
        for j, val in enumerate(row.values):
            c = cols[j]
            cls = "hi" if (highlight_col is not None and c == highlight_col) else ""
            cells.append(f'<td class="{cls}">{int(val)}</td>')
        body_rows.append("<tr>" + "".join(cells) + "</tr>")
    body_html = "\n".join(body_rows)
    return f"""{styles}
    <table class="nv">
      <caption>{title}</caption>
      <thead>{head_html}</thead>
      <tbody>{body_html}</tbody>
    </table>
    """

# --------------------------- UI (ipywidgets) ---------------------------

# Controls
scenario = w.RadioButtons(options=["Scenario A", "Scenario B", "Custom"], value="Scenario A", description="Scenario:")
rounds = w.BoundedIntText(value=20, min=1, max=200, step=1, description="Rounds:")
seed = w.Text(value="", description="Seed:")
student_id = w.Text(value="", description="Student ID:")
p_in = w.FloatText(value=12.0, description="p (price)")
c_in = w.FloatText(value=3.0, description="c (cost)")
h_in = w.FloatText(value=0.0, description="h (salvage)")
custom_box = w.HBox([p_in, c_in, h_in])
custom_box.layout.display = "none"  # hidden unless Custom chosen

# Order quantity input
order_q = w.BoundedIntText(
    value=DEMAND_VALUES[0],
    min=min(DEMAND_VALUES),
    max=max(DEMAND_VALUES),
    step=25,
    description="Order q:",
    disabled=True
)

start_btn = w.Button(description="Start / Reset", button_style="primary")
submit_btn = w.Button(description="Submit Order", button_style="", disabled=True)
download_btn = w.Button(description="Download CSV", button_style="", disabled=True)

# Outputs
metrics_out = w.Output()
prev_profit_out = w.Output()
prev_oc_out = w.Output()
ref_profit_out = w.Output()
round_msg_out = w.Output()
hist_table_out = w.Output()      # NEW: running demand table with mean/SD
hist_chart_out = w.Output()      # NEW: charts of historical demand
summary_out = w.Output()
results_table_out = w.Output()

# Game state
STATE = {
    "params": Params(12.0, 3.0, 0.0),
    "rounds_total": 10,
    "rng": random.Random(),
    "rng_seed": "",
    "results": [],
    "cum_profit": 0.0,
    "current_round": 1,
    "last_demand": None,
    "game_over": False
}

def render_metrics():
    metrics_out.clear_output()
    with metrics_out:
        print(f"Current Round: {min(STATE['current_round'], STATE['rounds_total'])} / {STATE['rounds_total']}")
        print(f"Cumulative Profit: {STATE['cum_profit']:.2f}")
        p = STATE["params"]
        print(f"Parameters: p={p.p}, c={p.c}, h={p.h}")

def render_prev_tables():
    prev_profit_out.clear_output()
    prev_oc_out.clear_output()
    if STATE["last_demand"] is None: return
    p = STATE["params"]
    with prev_profit_out:
        html = df_to_html_table(build_profit_table_df(p), "PROFIT TABLE (previous round context)", STATE["last_demand"])
        display(HTML(html))
    with prev_oc_out:
        html = df_to_html_table(build_opportunity_cost_table_df(p), "OPPORTUNITY COST TABLE (previous round context)", STATE["last_demand"])
        display(HTML(html))

def render_ref_profit():
    ref_profit_out.clear_output()
    with ref_profit_out:
        html = df_to_html_table(build_profit_table_df(STATE["params"]), "Reference: Profit Table (for your next decision)")
        display(HTML(html))

def render_round_msg(text):
    round_msg_out.clear_output()
    with round_msg_out:
        display(HTML(f"<div style='font-family:ui-sans-serif;'>{text}</div>"))

def render_historical_data():
    """NEW: shows the running demand table + mean/SD, and two charts
    (demand-by-round with running mean, and demand frequency) so students
    can estimate mu and sigma themselves."""
    hist_table_out.clear_output()
    hist_chart_out.clear_output()
    results = STATE["results"]

    with hist_table_out:
        display(HTML("<h4>📈 Historical Demand Data (use this to estimate μ and σ)</h4>"))
        if not results:
            print("No rounds played yet. Your observed demand history will build up here as you play.")
            return
        hist_df = pd.DataFrame({
            "Round": [r.round_num for r in results],
            "Demand": [r.demand_x for r in results],
            "Running Mean (μ̂)": [round(r.running_mean, 2) for r in results],
            "Running SD (σ̂)": [round(r.running_sd, 2) if r.running_sd is not None else None for r in results],
        })
        display(hist_df)
        last = results[-1]
        sd_text = f"{last.running_sd:.2f}" if last.running_sd is not None else "N/A (need ≥ 2 rounds)"
        print(f"\nCurrent estimate after {last.round_num} round(s):  μ̂ = {last.running_mean:.2f}   |   σ̂ = {sd_text}")

    if not results:
        with hist_chart_out:
            pass
        return

    demands = [r.demand_x for r in results]
    rounds_idx = [r.round_num for r in results]
    running_means = [r.running_mean for r in results]

    with hist_chart_out:
        fig, axes = plt.subplots(1, 2, figsize=(11, 4))

        # Left: demand realized per round + running mean line
        axes[0].plot(rounds_idx, demands, marker='o', linestyle='-', color="#4C72B0", label='Realized demand')
        axes[0].plot(rounds_idx, running_means, marker='', linestyle='--', color="#DD8452", label='Running mean (μ̂)')
        axes[0].set_xlabel("Round")
        axes[0].set_ylabel("Demand")
        axes[0].set_title("Demand by Round")
        axes[0].set_ylim(min(DEMAND_VALUES) - 20, max(DEMAND_VALUES) + 20)
        axes[0].legend()
        axes[0].grid(alpha=0.3)

        # Right: frequency of each demand value observed so far
        vals, counts = np.unique(demands, return_counts=True)
        all_vals = DEMAND_VALUES
        all_counts = [int(counts[list(vals).index(v)]) if v in vals else 0 for v in all_vals]
        axes[1].bar([str(v) for v in all_vals], all_counts, color="#55A868")
        axes[1].set_xlabel("Demand value")
        axes[1].set_ylabel("Frequency")
        axes[1].set_title("Demand Frequency (so far)")
        axes[1].grid(alpha=0.3, axis='y')

        plt.tight_layout()
        plt.show()

def reset_game(_=None):
    # Params
    if scenario.value == "Scenario A":
        STATE["params"] = Params(12.0, 3.0, 0.0)
    elif scenario.value == "Scenario B":
        STATE["params"] = Params(12.0, 9.0, 0.0)
    else:
        p = max(0.0, float(p_in.value))
        c = max(0.0, min(float(c_in.value), p))
        h = max(0.0, min(float(h_in.value), c))
        STATE["params"] = Params(p, c, h)

    # Take the current widget value as the new total
    STATE["rounds_total"] = int(rounds.value)

    STATE["rng_seed"] = seed.value.strip()
    try:
        STATE["rng"] = random.Random(int(STATE["rng_seed"])) if STATE["rng_seed"] else random.Random()
    except:
        STATE["rng"] = random.Random()

    STATE["results"] = []
    STATE["cum_profit"] = 0.0
    STATE["current_round"] = 1
    STATE["last_demand"] = None
    STATE["game_over"] = False

    # UI enable
    order_q.disabled = False
    submit_btn.disabled = False
    download_btn.disabled = True

    # Clear outputs
    prev_profit_out.clear_output()
    prev_oc_out.clear_output()
    ref_profit_out.clear_output()
    round_msg_out.clear_output()
    hist_table_out.clear_output()
    hist_chart_out.clear_output()
    summary_out.clear_output()
    results_table_out.clear_output()

    render_metrics()
    render_ref_profit()
    render_historical_data()
    render_round_msg("Game started. Choose your order quantity <b>q</b> and click <b>Submit Order</b>.")

def submit_order(_=None):
    if STATE["game_over"]:
        return
    q = order_q.value
    if q is None:
        render_round_msg("<span style='color:#b00;'>Please select an order quantity.</span>")
        return

    x = STATE["rng"].choice(DEMAND_VALUES)
    prof, sloss, hloss = profit_and_costs(q, x, STATE["params"])
    STATE["cum_profit"] += prof

    # running mean/SD of demand observed so far (including this round)
    prior_demands = [r.demand_x for r in STATE["results"]]
    all_demands = prior_demands + [x]
    running_mean = float(np.mean(all_demands))
    running_sd = float(np.std(all_demands, ddof=1)) if len(all_demands) > 1 else None

    rr = RoundResult(
        round_num=STATE["current_round"], order_q=q, demand_x=x,
        profit=prof, shortage_loss=sloss, salvage_loss=hloss,
        cum_profit=STATE["cum_profit"],
        running_mean=running_mean, running_sd=running_sd
    )
    STATE["results"].append(rr)
    STATE["last_demand"] = x
    STATE["current_round"] += 1
    if STATE["current_round"] > STATE["rounds_total"]:
        STATE["game_over"] = True

    render_metrics()
    render_prev_tables()
    render_ref_profit()
    render_historical_data()
    render_round_msg(
      f"📦 Your order: <b>{q}</b> &nbsp; | &nbsp; "
      f"✅ Demand realized: <b>{x}</b> &nbsp; | &nbsp; "
      f"Profit this round: <b>{prof:.2f}</b> "
      f"(Shortage loss: {sloss:.2f} | Salvage loss: {hloss:.2f})"
      )

    if STATE["game_over"]:
        # Summary + results + enable download
        summary_out.clear_output()
        with summary_out:
            p = STATE["params"]
            cr = critical_ratio(p)
            orders = [r.order_q for r in STATE["results"]]
            avg_q = sum(orders)/len(orders) if orders else 0.0
            final_mean = STATE["results"][-1].running_mean
            final_sd = STATE["results"][-1].running_sd
            total_profit = STATE["results"][-1].cum_profit if STATE["results"] else 0.0
            print("SUMMARY")
            print(f" - Your average order (qa): {avg_q:.2f}")
            print(f" - Your estimated demand mean (μ̂): {final_mean:.2f}")
            print(f" - Your estimated demand SD (σ̂): {final_sd:.2f}" if final_sd is not None else " - Your estimated demand SD (σ̂): N/A")
            print(f" - Critical ratio (CR = Cu/(Cu+Co)): {cr:.3f}")
            print(f" - Total profit over {STATE['rounds_total']} rounds: {total_profit:.2f}")
            print("\nNext step: use CR and a z-table (or the normal approximation Q* ≈ μ̂ + z_CR·σ̂)")
            print("to estimate your own Q*, then compute δq = qa − Q*. See the 'What You Need to")
            print("Submit' section above for the full deliverable.")

        df_results = pd.DataFrame([asdict(r) for r in STATE["results"]]).rename(columns={
            "round_num":"Round","order_q":"Order_q","demand_x":"Demand_x",
            "profit":"Profit","shortage_loss":"Shortage_Loss","salvage_loss":"Salvage_Loss",
            "cum_profit":"Cum_Profit","running_mean":"Running_Mean","running_sd":"Running_SD"
        })
        results_table_out.clear_output()
        with results_table_out:
            display(df_results)

        download_btn.disabled = False
        order_q.disabled = True
        submit_btn.disabled = True

def download_csv(_=None):
    if not STATE["results"]:
        return
    df_results = pd.DataFrame([asdict(r) for r in STATE["results"]]).rename(columns={
        "round_num":"Round","order_q":"Order_q","demand_x":"Demand_X",
        "profit":"Profit","shortage_loss":"Shortage_Loss","salvage_loss":"Salvage_Loss",
        "cum_profit":"Cum_Profit","running_mean":"Running_Mean","running_sd":"Running_SD"
    })
    sid = student_id.value.strip() or "student"
    ts = datetime.datetime.now().strftime("%Y%m%d-%H%M%S")
    fname = f"newsvendor_orders_{sid}_{ts}.csv"
    base_dir = "/content" if ON_COLAB else "."
    path = os.path.join(base_dir, fname)
    df_results.to_csv(path, index=False)
    if ON_COLAB:
        files.download(path)
    else:
        print(f"Saved to {path}")

def on_scenario_change(change):
    custom_box.layout.display = "" if change["new"] == "Custom" else "none"

def on_rounds_change(change):
    try:
        new_total = int(change["new"])
        if new_total >= 1:
            STATE["rounds_total"] = new_total
            render_metrics()
    except Exception:
        pass

scenario.observe(on_scenario_change, names="value")
rounds.observe(on_rounds_change, names="value")
start_btn.on_click(reset_game)
submit_btn.on_click(submit_order)
download_btn.on_click(download_csv)

# Layout
top_row = w.HBox([scenario, rounds, seed, student_id])
controls = w.VBox([
    top_row,
    custom_box,
    w.HBox([order_q]),
    w.HBox([start_btn, submit_btn, download_btn])
])
tables = w.VBox([
    w.HTML("<hr>"),
    w.HTML("<h3>Previous Round Context</h3>"),
    w.HBox([prev_profit_out, prev_oc_out]),
    w.HTML("<hr>"),
    w.HTML("<h3>Reference: Profit Table (for your next decision)</h3>"),
    ref_profit_out
])
history = w.VBox([
    w.HTML("<hr>"),
    hist_table_out,
    hist_chart_out
])

display(w.HTML("<h2>📰 Newsvendor Game (Colab)</h2>"))
display(w.HTML("<p>Pick a scenario, click <b>Start / Reset</b>, then choose an order quantity each round. "
               "You can change <b>Rounds</b> at any time; it updates right away.</p>"))
display(controls)
display(metrics_out)
display(round_msg_out)
display(tables)
display(history)
display(summary_out)
display(w.HTML("<h3>Your Orders & Outcomes</h3>"))
display(results_table_out)

# Auto start if you want:
# reset_game()
```
